# MFCC Visualization — Real vs Fake

Loads one **real** sample and one **fake** sample, extracts MFCCs, and plots both.

Purpose: verify that feature extraction works. This is **not** meant to prove
classification will succeed just because the plots look different.

```python
import librosa
import librosa.display
import matplotlib.pyplot as plt

SAMPLE_RATE = 16000
N_MFCC = 40

# Update these paths to point at one real and one fake sample
REAL_AUDIO_PATH = "real_sample.wav"
FAKE_AUDIO_PATH = "fake_sample.wav"
```

## Real audio

```python
real_audio, sr = librosa.load(REAL_AUDIO_PATH, sr=SAMPLE_RATE, mono=True)
real_mfcc = librosa.feature.mfcc(y=real_audio, sr=sr, n_mfcc=N_MFCC)

print("Real MFCC shape:", real_mfcc.shape)

plt.figure(figsize=(10, 5))
librosa.display.specshow(real_mfcc, x_axis="time", sr=sr)
plt.colorbar()
plt.title("MFCC — Real Audio")
plt.ylabel("MFCC Coefficients")
plt.xlabel("Time")
plt.tight_layout()
plt.show()
```

## Fake audio

```python
fake_audio, sr = librosa.load(FAKE_AUDIO_PATH, sr=SAMPLE_RATE, mono=True)
fake_mfcc = librosa.feature.mfcc(y=fake_audio, sr=sr, n_mfcc=N_MFCC)

print("Fake MFCC shape:", fake_mfcc.shape)

plt.figure(figsize=(10, 5))
librosa.display.specshow(fake_mfcc, x_axis="time", sr=sr)
plt.colorbar()
plt.title("MFCC — Fake Audio")
plt.ylabel("MFCC Coefficients")
plt.xlabel("Time")
plt.tight_layout()
plt.show()
```

## Note

This step only confirms the extraction pipeline runs correctly on both classes.
Any visual differences between the two plots are **not** by themselves evidence
that a classifier will separate real vs fake — that is what the modeling stage
will test.
